```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

import nltk
from nltk.corpus import wordnet as wn
from nltk.tokenize import word_tokenize

nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

def lesk(word, sentence):
    best_synset = None
    max_overlap = 0

    sentence_tokens = set(nltk.word_tokenize(sentence))

    for synset in wn.synsets(word):
        definition_tokens = set(nltk.word_tokenize(synset.definition()))
        examples = set(nltk.word_tokenize(' '.join(synset.examples())))

        overlap = len(definition_tokens.intersection(sentence_tokens)) + len(examples.intersection(sentence_tokens))

        if overlap > max_overlap:
            max_overlap = overlap
            best_synset = synset

    return best_synset
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

sentence = "I went to the river bank for a swim."
ambiguous_word = "bank"

synset = lesk(ambiguous_word, sentence)
print(f"The best synset for '{ambiguous_word}' in the sentence is: {synset.name()} -> {synset.definition()}")
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

dataset = [
    ("I went to the bank to deposit money.", "depository_financial_institution.n.01"),
    ("He has a bank of knowledge.", "bank.n.04")
]

correct_predictions = 0

for sentence, correct_sentence in dataset:
  predicted_synset = lesk("bank", sentence)
  if predicted_synset and predicted_synset.name() == correct_sentence:
    correct_predictions += 1

accuracy = correct_predictions / len(dataset)
print(f"Accuracy: {accuracy}")
```
