```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

documents = [

    "Python is a popular programming language for beginners.",

    "Data science uses statistics and programming to analyze data.",

    "Artificial intelligence helps machines perform smart tasks.",

    "The students attended a workshop on cloud computing.",

    "Cybersecurity protects computer systems from online threats."

]
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

for i, document in enumerate(documents):
    print(f"D{i + 1}: {document}")
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)
print(tfidf_matrix.shape)
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

print(tfidf_matrix)
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=["D1", "D2", "D3", "D4", "D5"]
)
tfidf_df.round(2)
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

print(vectorizer.get_feature_names_out())
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

len(vectorizer.get_feature_names_out())
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

similarity_matrix = cosine_similarity(tfidf_matrix)
similarity_matrix
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

similarity_df = pd.DataFrame(
similarity_matrix,
index=["D1", "D2", "D3", "D4", "D5"],
columns=["D1", "D2", "D3", "D4", "D5"]
)
similarity_df.round(3)
```

```python
# Student Details: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28
print("Name: Khan Yahya Shahid Ali | UIN: 221A048 | Roll No: 28")

highest_similarity = -1
most_similar_pair = None

for i in range(len(documents)):
    for j in range(i + 1, len(documents)):
        similarity = similarity_matrix[i][j]
        if similarity > highest_similarity:
            highest_similarity = similarity
            most_similar_pair = (i, j)

print(f"Most similar documents: D{most_similar_pair[0]+1} and D{most_similar_pair[1]+1}")
print(f"Similarity: {highest_similarity:.3f}")
```
